```python
import pandas as pd
```

```python
data=pd.read_excel("test.xlsx")
data
```

```python
predictions=data["prediction"].values
truth=data["truth"].values
truth
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import  matplotlib.pyplot  as plt 

cm=confusion_matrix(truth,predictions,labels=["fresh","rotten"])

plt.figure(figsize=(5,5))
sns.heatmap(cm,annot=True,xticklabels=["fresh","rotten"],yticklabels=["fresh","rotten"])

plt.show()

```

```python
from sklearn.metrics import accuracy_score 

accuracy_score(truth,predictions)
```

```python
from sklearn.metrics import precision_score

precision_score(truth,predictions,pos_label="fresh"),precision_score(truth,predictions,pos_label="rotten")
```
